### Compare MCM with early ARS

```python
output_dir = '../../results/compared_peaks'

ARS_early = pd.read_csv('../../data/references/ARS_early.bed',
                    names=['chr', 'start', 'end', 'center'], sep='\t', header=None)
MCM = pd.read_csv(f'{output_dir}/RDY368_G1_unTr_nT_nC_RDY412_G1_unTr_nT_nC/in_RDY412_G1_unTr_nT_nC_RDY368_G1_unTr_nT_nC_ex.bed', 
                    names=['chr', 'start', 'end', 'center'], sep='\t', header=None)
```

```python
origins = compare_pairwise({'ARS': ARS_early},
                            {'MCMs': MCM})
origins
```

```python
origins.to_csv(f'{output_dir}/eARS_with_MCMs_0bp.csv', sep='\t', header=True)
```

```python
origins = ARS_early.copy()

origins = compare_pairwise({'ARS': ARS_early},
                            {'MCMs': MCM},
                            include_neighbours=150)
origins
```
