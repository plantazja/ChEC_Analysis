```python
import sys, os
import pandas as pd
import numpy as np
import itertools

sys.path.append('../../scripts/2_comparative_analysis')

from compare import compare_pairwise
```

```python
# Load data
# Load coomon peaks for MCM2 and MCM4
MCM2_MCM4 = pd.read_csv('../../results/compared_peaks/RDY368_G1_unTr_nT_nC_RDY412_G1_unTr_nT_nC/in_RDY412_G1_unTr_nT_nC_RDY368_G1_unTr_nT_nC_ex.bed', names=['chr', 'start', 'end', 'center'], sep='\t')

# Load ARS SGD
ARS_SGD = pd.read_csv('../../data/references/ARS_late.bed', names=['chr', 'start', 'end', 'center'], sep='\t')

# Load ARS early and late
ARS_early_late = pd.read_csv('../../data/references/SGD_ARS_annot_only_early_late.csv', sep=',')
ARS_early = ARS_early_late[ARS_early_late['Early_late_Yabuki2002'] == 'early'][['chr', 'ARS_left', 'ARS_right', 'ARS_center']]
ARS_early.columns = ['chr', 'start', 'end', 'center']

ARS_late = ARS_early_late[ARS_early_late['Early_late_Yabuki2002'] == 'late'][['chr', 'ARS_left', 'ARS_right', 'ARS_center']]
ARS_late.columns = ['chr', 'start', 'end', 'center']
```

```python
MCM2_MCM4
```

```python
def check_match(row, factor):
    mask = (factor['chr'] == row['chr']) & \
           (row['center'] + 75 >= factor['start']) & \
           (row['center'] - 75 <= factor['end'])
    return mask.any()
```

```python
MCM = pd.DataFrame()
MCM['chr'] = MCM2_MCM4['chr']
MCM['center'] = MCM2_MCM4['center']

MCM['ARS_SGD'] = MCM.apply(check_match, axis=1, factor=ARS_SGD).astype(int)
MCM['ARS_early'] = MCM.apply(check_match, axis=1, factor=ARS_early).astype(int)
MCM['ARS_late'] = MCM.apply(check_match, axis=1, factor=ARS_late).astype(int)
MCM
```

```python
X = compare_pairwise({'MCM': MCM2_MCM4},
                    {'ARS_early': ARS_early})
X
```
